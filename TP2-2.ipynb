## Ejercicios Teóricos

1. ¿Qué diferencia hay entre una estrategia de búsqueda Informada y una estrategia de búsqueda No Informada?

Búsqueda no informada: el algoritmo no sabe qué camino es mejor. Solo explora los caminos disponibles hasta encontrar el objetivo.
Ejemplos: búsqueda en anchura, profundidad y costo uniforme.
Búsqueda informada: el algoritmo tiene información extra (heurística) que le ayuda a decidir qué camino parece más conveniente para llegar al objetivo.
Ejemplos: búsqueda voraz

2. ¿Qué es una heurística y para qué sirve?

La función heurística h(n) es una estimación de cuánto falta para llegar al objetivo desde un nodo.
Sirve para que el algoritmo pueda elegir qué camino parece más conveniente y no buscar a ciegas.
Ejemplo: si queremos llegar de una ciudad a otra, podemos usar la distancia en línea recta como estimación de cuánto falta.
Sirve para:
Guía la búsqueda: indica qué nodos parecen más cercanos al objetivo.
Hace la búsqueda más eficiente: evita explorar caminos que parecen poco convenientes.
En A* se combina con el costo recorrido:
f(n)=g(n)+h(n)
donde:
g(n) = costo que ya recorrimos.
h(n) = costo que estimamos que falta.

3. ¿Es posible que un algoritmo de búsqueda no tenga solución?

Sí, un algoritmo de búsqueda puede no encontrar una solución. Esto puede pasar por dos motivos:
1. La solución realmente no existe
No hay ningún camino que lleve desde el estado inicial hasta el objetivo.
En este caso, un algoritmo completo explora todas las posibilidades y puede determinar que no existe solución.

2. La solución existe, pero el algoritmo no la encuentra
Esto ocurre cuando el algoritmo no es completo.
Por ejemplo:
Búsqueda en profundidad: puede quedarse atrapada en un camino infinito.
Búsqueda de profundidad limitada: puede no llegar al objetivo si está más allá del límite establecido.
Búsqueda voraz: puede elegir un camino incorrecto y no volver para probar otras opciones.

**Secuencia total del algoritmo** (concatenando las iteraciones):

`A, D, F, I` → `A, D, H, J, F, C, E, I` → `A, D, H, P, O, J, K, F, C, Z, W, E, I` → `A, D, H, P, O, J, K, B, F, C, Z, W, E, I`

Coordenadas para la Distancia de Manhattan ($h = \vert{}x_1 - x_2\vert{} + \vert{}y_1 - y_2\vert{}$):Objetivo (F): (3, 2)Coordenadas: A(0,1), B(0,2), C(1,1), D(1,2), E(1,3), G(2,0), I(2,1), W(2,2), K(2,3), M(2,4), N(2,5), P(3,0), Q(3,1), R(3,2), T(3,3), F(3,4).Paredes rojas (bloquean el paso):Entre C y DEntre D y EEntre W y REntre T y F

```python
import heapq

# Coordenadas (fila, columna) para Distancia de Manhattan a F (3, 4)
pos = {
    'A': (0, 1), 'B': (0, 2),
    'C': (1, 1), 'D': (1, 2), 'E': (1, 3),
    'G': (2, 0), 'I': (2, 1), 'W': (2, 2), 'K': (2, 3), 'M': (2, 4), 'N': (2, 5),
    'P': (3, 0), 'Q': (3, 1), 'R': (3, 2), 'T': (3, 3), 'F': (3, 4)
}

# Grafo no dirigido con las paredes rojas aplicadas correctamente
graph = {
    'A': ['B', 'C'],
    'B': ['A'],
    'C': ['A', 'I'],
    'D': [],  # Bloqueado por paredes (C-D y D-E)
    'E': ['N'],
    'G': ['I', 'P'],
    'I': ['C', 'G', 'Q', 'W'],
    'W': ['I', 'K'],
    'K': ['M', 'T', 'W'],
    'M': ['K', 'N'],
    'N': ['E', 'M'],
    'P': ['G', 'Q'],
    'Q': ['I', 'P', 'R'],
    'R': ['Q'],        # Pared entre R y F (no hay paso directo)
    'T': ['F', 'K'],   # Paso habilitado entre T y F
    'F': ['T']         # F conectada con T
}

def heuristic(node):
    x1, y1 = pos[node]
    x2, y2 = pos['F']
    return abs(x1 - x2) + abs(y1 - y2)

def step_cost(node):
    return 30 if node == 'W' else 1

# 5.1 Primero en Profundidad (DFS)
def dfs(start='I', goal='F'):
    stack = [(start, [start])]
    visited = set()
    
    while stack:
        node, path = stack.pop()
        if node == goal:
            return path
        if node not in visited:
            visited.add(node)
            # Orden inverso para que en el stack el pop() saque en orden alfabético
            neighbors = sorted([n for n in graph[node] if n not in visited], reverse=True)
            for neighbor in neighbors:
                stack.append((neighbor, path + [neighbor]))
    return None

# 5.2 Búsqueda Avara (Greedy Best-First)
def greedy_search(start='I', goal='F'):
    pq = [(heuristic(start), start, [start])]
    visited = set()
    
    while pq:
        h, node, path = heapq.heappop(pq)
        if node == goal:
            return path
        if node not in visited:
            visited.add(node)
            neighbors = sorted([n for n in graph[node] if n not in visited])
            for neighbor in neighbors:
                heapq.heappush(pq, (heuristic(neighbor), neighbor, path + [neighbor]))
    return None

# 5.3 Búsqueda A*
def a_star_search(start='I', goal='F'):
    pq = [(heuristic(start), 0, start, [start])]
    visited = {}
    
    while pq:
        f, g, node, path = heapq.heappop(pq)
        if node == goal:
            return path, g
        
        if node not in visited or g < visited[node]:
            visited[node] = g
            neighbors = sorted(graph[node])
            for neighbor in neighbors:
                g_new = g + step_cost(neighbor)
                f_new = g_new + heuristic(neighbor)
                heapq.heappush(pq, (f_new, g_new, neighbor, path + [neighbor]))
    return None, float('inf')

# Ejecución y salida limpia
res_dfs = dfs()
print("5.1 Camino DFS:  ", " -> ".join(res_dfs) if res_dfs else "No se encontró camino")

res_greedy = greedy_search()
print("5.2 Camino Avara:", " -> ".join(res_greedy) if res_greedy else "No se encontró camino")

path_a_star, cost_a_star = a_star_search()
print("5.3 Camino A*:   ", " -> ".join(path_a_star) if path_a_star else "No se encontró camino", f"| Costo total: {cost_a_star}")
```

```python
"""
Agente de busqueda A* sobre una grilla 2D con una pared en forma de "L".

El archivo esta dividido en dos secciones bien separadas:
    1) NUCLEO ALGORITMICO  -> formulacion del problema + algoritmo A*
       (no contiene absolutamente nada de dibujo/matplotlib)
    2) VISUALIZACION       -> toma el resultado del algoritmo y lo dibuja

Se sigue el esquema visto en clase para resolver un problema de busqueda:
    - Formulacion del objetivo / estado inicial / estado objetivo
    - Formulacion del problema (acciones, funcion sucesor, costo de camino)
    - Busqueda (algoritmo A*, con funcion heuristica y criterio de desempate)
    - Ejecucion (reconstruccion e impresion del camino)
"""

import heapq
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import matplotlib.animation as animation


# =====================================================================
# SECCION 1: NUCLEO ALGORITMICO (sin nada de visualizacion)
# =====================================================================

# ---------------------------------------------------------------------
# 1.a) FORMULACION DEL PROBLEMA: estados, estado inicial y estado objetivo
# ---------------------------------------------------------------------
# La grilla se referencia como (columna, fila), con fila 0 arriba y
# columna 0 a la izquierda.
#
# Coordenadas de la grilla, la pared, el inicio (I) y el fin (F) dadas
# explicitamente para este ejercicio.

ANCHO = 17   # cantidad de columnas de la grilla
ALTO = 16    # cantidad de filas de la grilla

# Estado inicial: la casilla "I".
ESTADO_INICIAL = (4, 4)

# Estado objetivo (test objetivo: estado == ESTADO_OBJETIVO): casilla "F".
ESTADO_OBJETIVO = (11, 12)

# Pared: conjunto de casillas bloqueadas / no transitables.
# - Quiebre superior (dos celdas sueltas cerca de la entrada al bolsillo).
# - Barra vertical: columna x=9, filas y de 5 a 14.
# - Barra horizontal: fila y=14, columnas x de 9 a 12.
PARED = set()
PARED.add((11, 3))
PARED.add((10, 4))
for fila in range(5, 15):
    PARED.add((9, fila))
for col in range(9, 13):
    PARED.add((col, 14))


# ---------------------------------------------------------------------
# 1.b) FUNCION SUCESOR: que acciones/estados son alcanzables desde uno dado
# ---------------------------------------------------------------------
# Movimiento en 4 direcciones (arriba, abajo, izquierda, derecha).
# Un sucesor es valido si esta dentro de la grilla y no es una casilla
# de pared.

ACCIONES = [(0, -1), (0, 1), (-1, 0), (1, 0)]  # arriba, abajo, izq, der


def sucesores(estado):
    x, y = estado
    for dx, dy in ACCIONES:
        nx, ny = x + dx, y + dy
        dentro_de_grilla = 0 <= nx < ANCHO and 0 <= ny < ALTO
        if dentro_de_grilla and (nx, ny) not in PARED:
            yield (nx, ny)


# ---------------------------------------------------------------------
# 1.c) COSTO DE CAMINO: cada paso (accion) cuesta 1
# ---------------------------------------------------------------------
COSTO_PASO = 1


# ---------------------------------------------------------------------
# 1.d) FUNCION HEURISTICA h(n): distancia Manhattan al objetivo
# ---------------------------------------------------------------------
# Es admisible (nunca sobreestima el costo real) porque en una grilla de
# movimiento 4-direccional el minimo numero de pasos posible entre dos
# casillas, ignorando obstaculos, es exactamente |dx| + |dy|. Al ser
# admisible, A* garantiza encontrar la solucion optima.

def h(estado):
    x, y = estado
    xf, yf = ESTADO_OBJETIVO
    return abs(x - xf) + abs(y - yf)


# ---------------------------------------------------------------------
# 1.e) NODO DE BUSQUEDA
# ---------------------------------------------------------------------
# Cada nodo guarda su propio camino recorrido (visitados_en_rama) en vez
# de usar una unica lista cerrada global. Esto implementa la regla
# adicional pedida: "no revisitar una casilla ya seleccionada dentro de
# la MISMA rama", sin que eso bloquee a otras ramas de la busqueda que
# podrian necesitar pasar por esa misma casilla.

class Nodo:
    __slots__ = ("estado", "padre", "g", "h", "f", "visitados_en_rama")

    def __init__(self, estado, padre, g, visitados_en_rama):
        self.estado = estado
        self.padre = padre
        self.g = g                       # costo acumulado desde la raiz
        self.h = h(estado)                # estimacion heuristica al objetivo
        self.f = self.g + self.h          # funcion de evaluacion f(n) = g(n) + h(n)
        self.visitados_en_rama = visitados_en_rama  # frozenset propio de esta rama


def reconstruir_camino(nodo):
    camino = []
    actual = nodo
    while actual is not None:
        camino.append(actual.estado)
        actual = actual.padre
    camino.reverse()
    return camino


# ---------------------------------------------------------------------
# 1.f) ALGORITMO A*
# ---------------------------------------------------------------------
# La frontera es una cola de prioridad (min-heap) ordenada por f(n).
# Como heapq no puede comparar Nodos directamente, cada entrada de la
# frontera es una tupla (f, h, orden_de_insercion, nodo):
#   - se ordena primero por f(n)              -> criterio principal de A*
#   - en caso de empate en f(n), por h(n)     -> criterio de desempate elegido
#   - si tambien empata h(n), por orden de insercion -> desempate final
#     (equivalente a comportamiento FIFO entre nodos totalmente empatados)
#
# REGLA "NO REVISITAR DENTRO DE LA MISMA RAMA":
# Cada nodo mantiene su propio Nodo.visitados_en_rama, y un sucesor que ya
# fue pisado en ESE camino especifico se descarta. Esto evita que una rama
# individual entre en un ciclo infinito (ir y volver sobre sus propios pasos),
# y es intencionalmente independiente de lo que hicieron otras ramas, tal
# como se pidio.
#
# PODA POR DOMINANCIA (necesaria en una grilla, a diferencia del mapa de
# ciudades disperso de la teoria): en una grilla abierta existen MUCHISIMOS
# caminos distintos con el mismo costo entre dos celdas (ej. moverse
# derecha-derecha-abajo vs abajo-derecha-derecha llega al mismo lugar con el
# mismo costo). Si cada uno de esos caminos equivalentes generara su propia
# rama sin ningun tipo de control global, la cantidad de nodos crece de forma
# combinatoria (explota) y la busqueda no termina en tiempo razonable.
# Por eso se usa un diccionario global `mejor_g_conocido`: si a un estado ya
# se llego antes con un costo g menor o igual, la nueva rama que llega a ese
# mismo estado con g mayor o igual se descarta (no puede mejorar la solucion,
# y con heuristica admisible/consistente esto no afecta la optimalidad).
# Notar que esto NO es lo mismo que la regla de "no revisitar en la misma
# rama": aquella evita ciclos dentro de un camino; esta poda evita reexplorar
# trabajo redundante entre ramas distintas que ya sabemos que no puede ganar.
#
# ORDEN DE EXPANSION: ademas del set `explorados_global` (usado solo para
# consultas rapidas de pertenencia), se guarda una LISTA `orden_expansion`
# con los estados en el mismo orden exacto en que A* los fue expandiendo.
# Esta lista es pura salida de datos del algoritmo (igual que `camino`), no
# contiene ninguna logica de dibujo; la usa la Seccion 2 para animar la
# exploracion en el mismo orden real en que el algoritmo trabajo.

def busqueda_a_estrella(estado_inicial, estado_objetivo):
    contador_insercion = 0

    raiz = Nodo(estado_inicial, padre=None, g=0,
                visitados_en_rama=frozenset({estado_inicial}))

    frontera = []
    heapq.heappush(frontera, (raiz.f, raiz.h, contador_insercion, raiz))
    contador_insercion += 1

    explorados_global = set()      # para consultas rapidas de pertenencia
    orden_expansion = []           # orden real de expansion, para animar despues
    mejor_g_conocido = {}          # poda por dominancia: estado -> mejor g visto

    while frontera:
        _, _, _, nodo_actual = heapq.heappop(frontera)

        # Si para este estado ya conocemos un camino igual o mejor, se descarta
        # (puede haber quedado obsoleto en la frontera por venir de otra rama).
        if nodo_actual.g > mejor_g_conocido.get(nodo_actual.estado, float("inf")):
            continue

        # Test objetivo
        if nodo_actual.estado == estado_objetivo:
            camino = reconstruir_camino(nodo_actual)
            return camino, nodo_actual.g, orden_expansion

        explorados_global.add(nodo_actual.estado)
        orden_expansion.append(nodo_actual.estado)

        # Expansion: generar sucesores y filtrar los ya visitados EN ESTA RAMA
        for sucesor in sucesores(nodo_actual.estado):
            if sucesor in nodo_actual.visitados_en_rama:
                continue  # evita ciclos dentro de la misma rama

            nuevo_g = nodo_actual.g + COSTO_PASO

            # Poda por dominancia: si ya existe un camino a `sucesor` con
            # costo menor o igual, esta rama no puede aportar una mejor
            # solucion final, asi que no se genera el nodo hijo.
            if nuevo_g >= mejor_g_conocido.get(sucesor, float("inf")):
                continue
            mejor_g_conocido[sucesor] = nuevo_g

            nuevos_visitados = nodo_actual.visitados_en_rama | {sucesor}
            hijo = Nodo(sucesor, padre=nodo_actual, g=nuevo_g,
                        visitados_en_rama=nuevos_visitados)

            heapq.heappush(frontera, (hijo.f, hijo.h, contador_insercion, hijo))
            contador_insercion += 1

    return None, None, orden_expansion  # no se encontro solucion


# =====================================================================
# SECCION 2: VISUALIZACION ANIMADA (matplotlib) -- no toma NINGUNA decision
# del algoritmo, solo reproduce en pantalla los resultados ya calculados
# arriba (camino y orden_expansion).
# =====================================================================

# Velocidad de la animacion: milisegundos entre cada casilla que se ilumina.
INTERVALO_MS = 60

# Color claro usado para marcar una casilla ya expandida por el algoritmo.
COLOR_EXPLORADO = "#dcdcdc"
COLOR_CAMINO = "#ff9900"


def _dibujar_base(ax, ancho, alto, pared, inicio, fin):
    """Dibuja los elementos fijos de la grilla (pared, I, F, cuadricula)."""
    for (x, y) in pared:
        ax.add_patch(patches.Rectangle((x, alto - 1 - y), 1, 1,
                                        facecolor="black", edgecolor="none"))

    ix, iy = inicio
    fx, fy = fin
    ax.add_patch(patches.Rectangle((ix, alto - 1 - iy), 1, 1,
                                    facecolor="#3355ff", edgecolor="none"))
    ax.text(ix + 0.5, alto - 1 - iy + 0.5, "I", ha="center", va="center",
            color="white", fontweight="bold", zorder=5)

    ax.add_patch(patches.Rectangle((fx, alto - 1 - fy), 1, 1,
                                    facecolor="#ffee00", edgecolor="none"))
    ax.text(fx + 0.5, alto - 1 - fy + 0.5, "F", ha="center", va="center",
            color="black", fontweight="bold", zorder=5)

    ax.set_xlim(0, ancho)
    ax.set_ylim(0, alto)
    ax.set_xticks(range(ancho + 1))
    ax.set_yticks(range(alto + 1))
    ax.grid(True, color="black", linewidth=0.5)
    ax.set_xticklabels([])
    ax.set_yticklabels([])
    ax.set_aspect("equal")


def animar_busqueda(ancho, alto, pared, inicio, fin, orden_expansion, camino):
    """
    Anima, casilla por casilla, el orden real en que A* fue expandiendo
    nodos (colores claros). Cuando termina de mostrar toda la exploracion,
    resalta de una sola vez el camino final encontrado (naranja).
    """
    fig, ax = plt.subplots(figsize=(ancho / 2, alto / 2))
    _dibujar_base(ax, ancho, alto, pared, inicio, fin)
    ax.set_title("Busqueda A*: explorando...")

    # Un frame por casilla expandida, mas un frame extra al final para
    # dibujar el camino solucion completo.
    total_frames = len(orden_expansion) + 1

    def actualizar(frame):
        if frame < len(orden_expansion):
            # Se fue iluminando de a una casilla por frame (colores claros)
            x, y = orden_expansion[frame]
            ax.add_patch(patches.Rectangle((x, alto - 1 - y), 1, 1,
                                            facecolor=COLOR_EXPLORADO,
                                            edgecolor="none"))
        else:
            # Ultimo frame: se termino de explorar, se muestra el camino
            # final completo de una sola vez.
            if camino:
                for (cx, cy) in camino:
                    ax.add_patch(patches.Rectangle((cx, alto - 1 - cy), 1, 1,
                                                     facecolor=COLOR_CAMINO,
                                                     edgecolor="none"))
                ax.set_title("Busqueda A*: camino encontrado")
            else:
                ax.set_title("Busqueda A*: no se encontro solucion")
        return []

    anim = animation.FuncAnimation(fig, actualizar, frames=total_frames,
                                    interval=INTERVALO_MS, repeat=False)

    plt.tight_layout()
    plt.show()
    return anim  # se retorna para que la animacion no sea recolectada por el GC


# =====================================================================
# MAIN
# =====================================================================

def main():
    camino, costo_total, orden_expansion = busqueda_a_estrella(ESTADO_INICIAL, ESTADO_OBJETIVO)

    if camino is None:
        print("No se encontro un camino entre el inicio y el objetivo.")
        return

    print("Camino encontrado:")
    print(" -> ".join(str(c) for c in camino))
    print(f"Costo total del camino (numero de pasos): {costo_total}")
    print(f"Casillas expandidas durante la busqueda: {len(orden_expansion)}")

    animar_busqueda(ANCHO, ALTO, PARED, ESTADO_INICIAL, ESTADO_OBJETIVO,
                     orden_expansion, camino)


if __name__ == "__main__":
    main()
```
