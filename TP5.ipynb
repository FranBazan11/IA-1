1. ¿Cuáles son los tres axiomas de Kolmogorov?

Los tres axiomas de Kolmogorov definen formalmente qué es una probabilidad. Sea S el espacio muestral y A, B eventos (subconjuntos de S):

**1. No negatividad**
P(A) ≥ 0, para todo evento A

La probabilidad nunca puede ser negativa.

**2. Normalización (certeza)**
P(S) = 1

La probabilidad del espacio muestral completo (que ocurra "algo") es 1.

**3. Aditividad numerable (para eventos mutuamente excluyentes)**
Si A ∩ B = ∅ (A y B son disjuntos, no pueden ocurrir ambos a la vez), entonces:

P(A ∪ B) = P(A) + P(B)

Esto se extiende a una cantidad numerable de eventos disjuntos A₁, A₂, A₃, ...:

P(A₁ ∪ A₂ ∪ A₃ ∪ ...) = P(A₁) + P(A₂) + P(A₃) + ...

---

**De estos tres axiomas se derivan todas las demás propiedades**, como por ejemplo:

- P(∅) = 0
- P(¬A) = 1 − P(A)
- Si A ⊆ B, entonces P(A) ≤ P(B)
- P(A ∪ B) = P(A) + P(B) − P(A ∩ B) (regla general, para eventos no necesariamente disjuntos)
- 0 ≤ P(A) ≤ 1



2. Una fábrica de clavos dispone de 2 máquinas que elaboran el 30% y 70% de los clavos que producen respectivamente. El porcentaje de clavos defectuosos de cada máquina es del 2% y 3%, respectivamente. Si se selecciona al azar un clavo de la producción y este fue defectuoso, ¿cuál es la probabilidad de que haya sido fabricado por la máquina?

## Resolución

### Datos

- P(M₁) = 0,30
- P(M₂) = 0,70
- P(D|M₁) = 0,02
- P(D|M₂) = 0,03

### Teorema de Bayes (ecuación general)

P(Mᵢ|D) = [P(D|Mᵢ) · P(Mᵢ)] / P(D)

### Aplicando nuestros parámetros (para M₁)

P(M₁|D) = [P(D|M₁) · P(M₁)] / P(D)

Acá aparece P(D) en el denominador, que no es un dato directo — hay que calcularlo. Para eso se aplica el **teorema de probabilidad total**, ya que D puede ocurrir a través de M₁ o de M₂ (eventos mutuamente excluyentes y exhaustivos).

### Teorema de probabilidad total (ecuación general)

P(D) = P(D|M₁) · P(M₁) + P(D|M₂) · P(M₂)

### Reemplazando y calculando P(D)

P(D) = (0,02)(0,30) + (0,03)(0,70)
P(D) = 0,006 + 0,021
P(D) = 0,027

### Volviendo a Bayes — reemplazando y calculando

P(M₁|D) = (0,02 · 0,30) / 0,027 = 0,006 / 0,027 ≈ 0,2222

**P(M₁|D) ≈ 22,22%**

---

Para P(M₂|D), se resuelve igual, con el mismo denominador:

P(M₂|D) = (0,03 · 0,70) / 0,027 = 0,021 / 0,027 ≈ 0,7778

**P(M₂|D) ≈ 77,78%**

5. Diseñar un controlador difuso para jugar al Pong. El juego consiste en una pelota que rebota dentro de una cancha rectangular y palas controladas por el sistema difuso o el jugador. El entorno debe cumplir las siguientes reglas:

* La cancha posee dimensiones fijas.
* La pelota rebota en los bordes superior e inferior. Los bordes laterales funcionan como un arco y cuando la pelota atraviesa un borde suma un punto al lado opuesto, sin importar quién toca la pelota por última vez.
* La velocidad de la pelota debe ir incrementando con cada rebot en una pala o pared. Marcar un gol restablece los valores de velocidad a su configuración original.
* Cada pala únicamente puede desplazarse verticalmente y con velocidad fija durante toda la ejecución.
* Cada pala tiene una altura fija.
* Es necesario definir un puntaje mínimo que de el juego por finalizado.

La variable de salida sobre la que actúa el controlador es la **velocidad de la pala**. Las variables de entrada quedan a definir por ustedes.

Como mínimo deben hacer un modo de "un jugador" (ustedes vs. el agente), pero pueden incorporar un modo de "dos jugadores" que consista en utilizar dos agentes con variaciones en las reglas del motor de inferencia, mecanismos de fuzzificación o desfuzzificación de la información y/o las funciones de pertenencia de cada grupo.

```python
import pygame
import random
import sys
import math #para darle forma de flor

# ==========================================
# CONFIGURACIÓN DEL JUEGO Y COLORES
# ==========================================
WIDTH, HEIGHT = 800, 600
FPS = 60
WHITE = (255, 255, 255)
BLACK = (0, 0, 0)
GRAY = (100, 100, 100)
CELESTE = (119, 181, 254)
ROSADO = (255, 182, 193)  
AMARILLO = (255, 255, 180) 

MAX_SCORE = 5 # Puntaje máximo configurable

# ==========================================
# CONTROLADOR DIFUSO (Lógica implementada desde cero)
# ==========================================
class ControladorDifuso:
    def __init__(self):
        pass
    
    # Funciones de membresía base
    def trapecio(self, x, a, b, c, d):
        if x <= a or x >= d: return 0.0
        if a < x <= b: return (x - a) / (b - a) if b != a else 1.0
        if b < x <= c: return 1.0
        if c < x < d: return (d - x) / (d - c) if d != c else 1.0
        return 0.0

    def triangulo(self, x, a, b, c):
        if x <= a or x >= c: return 0.0
        if a < x <= b: return (x - a) / (b - a) if b != a else 1.0
        if b < x < c: return (c - x) / (c - b) if c != b else 1.0
        return 0.0

    def calcular_velocidad(self, ball_x, ball_y, paddle_x, paddle_y):
        # 1. Variables de Entrada
        error_y = ball_y - paddle_y   # Positivo = bola abajo, Negativo = bola arriba
        dist_x = abs(ball_x - paddle_x) # Distancia horizontal entre bola y pala

        # 2. Fuzzificación
        # Conjuntos para Error Y
        err_arriba = self.trapecio(error_y, -600, -600, -30, -10)
        err_centro = self.triangulo(error_y, -20, 0, 20)
        err_abajo = self.trapecio(error_y, 10, 30, 600, 600)

        # Conjuntos para Distancia X
        dist_cerca = self.trapecio(dist_x, 0, 0, 200, 400)
        dist_lejos = self.trapecio(dist_x, 200, 400, 800, 800)

        # 3. Base de Reglas e Inferencia (Operador AND = min)
        reglas = [
            (min(err_arriba, dist_cerca), -9),
            (min(err_arriba, dist_lejos), -4),
            (err_centro, 0),
            (min(err_abajo, dist_lejos), 4),
            (min(err_abajo, dist_cerca), 9)
        ]

        # 4. Desfuzzificación (Método de Centroide de Singletons - Promedio Ponderado)
        numerador = sum(grado * salida for grado, salida in reglas)
        denominador = sum(grado for grado, salida in reglas)

        if denominador == 0:
            return 0
        return numerador / denominador

# ==========================================
# CLASES DEL JUEGO
# ==========================================
class Pala:
    def __init__(self, x, y):
        # La pala es un rectángulo de 15x100 píxeles
        self.rect = pygame.Rect(x, y, 15, 100)
        self.speed = 15

    def mover_usuario(self, keys):
        if keys[pygame.K_UP] and self.rect.top > 0:
            self.rect.y -= self.speed
        if keys[pygame.K_DOWN] and self.rect.bottom < HEIGHT:
            self.rect.y += self.speed

    def mover_difuso(self, fuzzy_speed):
        self.rect.y += fuzzy_speed
        # Mantener la pala dentro de la pantalla
        if self.rect.top < 0:
            self.rect.top = 0
        if self.rect.bottom > HEIGHT:
            self.rect.bottom = HEIGHT

    def draw(self, surface):
    
        pygame.draw.rect(surface, CELESTE, self.rect)

class Pelota:
    def __init__(self):
       
        self.rect = pygame.Rect(WIDTH//2 - 20, HEIGHT//2 - 20, 40, 40)
        self.base_speed = 5.0
        self.speed_x = self.base_speed * random.choice([1, -1])
        self.speed_y = self.base_speed * random.choice([1, -1])

    def mover(self):
        self.rect.x += self.speed_x
        self.rect.y += self.speed_y

    def aumentar_velocidad(self):
        incremento = 1.05
        max_speed = 12.0
        if abs(self.speed_x) < max_speed:
            self.speed_x *= incremento
        if abs(self.speed_y) < max_speed:
            self.speed_y *= incremento

    def rebotar_pared_y(self):
        if self.rect.top <= 0 or self.rect.bottom >= HEIGHT:
            self.speed_y *= -1
            self.aumentar_velocidad()

    def reset(self):
        self.rect.center = (WIDTH//2, HEIGHT//2)
        self.speed_x = self.base_speed * random.choice([1, -1])
        self.speed_y = self.base_speed * random.choice([1, -1])

    def draw(self, surface):
        center_x, center_y = self.rect.center
        
        #Forma FLOR
        center_radius = 10     
        petal_radius = 12     
        dist_petalos = 14      
        num_petals = 5         

        
        for i in range(num_petals):
            angle = (360 / num_petals) * i
            radian_angle = math.radians(angle)
            
            # Posición de cada pétalo
            petal_x = center_x + dist_petalos * math.cos(radian_angle)
            petal_y = center_y + dist_petalos * math.sin(radian_angle)
            
            pygame.draw.circle(surface, ROSADO, (int(petal_x), int(petal_y)), petal_radius)

        
        pygame.draw.circle(surface, CELESTE, (center_x, center_y), center_radius)
# ==========================================
# BUCLE PRINCIPAL DEL JUEGO
# ==========================================
def main():
    pygame.init()
    screen = pygame.display.set_mode((WIDTH, HEIGHT))
    pygame.display.set_caption("Pong - Control Difuso vs Flor")
    clock = pygame.time.Clock()
    font = pygame.font.Font(None, 74)
    small_font = pygame.font.Font(None, 36)

    pala_jugador = Pala(30, HEIGHT//2 - 50)
    pala_ia = Pala(WIDTH - 45, HEIGHT//2 - 50)
    pelota = Pelota()
    controlador = ControladorDifuso()

    score_jugador = 0
    score_ia = 0
    juego_terminado = False
    ganador = ""

    while True:
        # 1. Manejo de Eventos
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                pygame.quit()
                sys.exit()
            if event.type == pygame.KEYDOWN:
                if event.key == pygame.K_SPACE and juego_terminado:
                    # Reiniciar partida
                    score_jugador = 0
                    score_ia = 0
                    juego_terminado = False
                    pelota.reset()

        if not juego_terminado:
            # 2. Lógica del Juego
            keys = pygame.key.get_pressed()
            pala_jugador.mover_usuario(keys)
            
            # Movimiento de la IA Difusa
            vel_difusa = controlador.calcular_velocidad(
                pelota.rect.centerx, pelota.rect.centery, 
                pala_ia.rect.centerx, pala_ia.rect.centery
            )
            pala_ia.mover_difuso(vel_difusa)

            pelota.mover()
            pelota.rebotar_pared_y()

            # Colisiones con las palas
            if pelota.rect.colliderect(pala_jugador.rect) and pelota.speed_x < 0:
                pelota.speed_x *= -1
                pelota.aumentar_velocidad()
            
            if pelota.rect.colliderect(pala_ia.rect) and pelota.speed_x > 0:
                pelota.speed_x *= -1
                pelota.aumentar_velocidad()

            # Sistema de puntuación
            if pelota.rect.left <= 0:
                score_ia += 1
                pelota.reset()
            elif pelota.rect.right >= WIDTH:
                score_jugador += 1
                pelota.reset()

            # Chequear fin de juego
            if score_jugador >= MAX_SCORE:
                juego_terminado = True
                ganador = "¡GANASTE!"
            elif score_ia >= MAX_SCORE:
                juego_terminado = True
                ganador = "GANÓ LA IA DIFUSA"

        # 3. Dibujo en Pantalla
        screen.fill(AMARILLO)
        pygame.draw.aaline(screen, GRAY, (WIDTH//2, 0), (WIDTH//2, HEIGHT))

        pala_jugador.draw(screen)
        pala_ia.draw(screen)
        pelota.draw(screen)

        # Textos de puntuación con ROSADO_BEBE
        texto_jug = font.render(str(score_jugador), True, ROSADO)
        texto_ia = font.render(str(score_ia), True, ROSADO)
        screen.blit(texto_jug, (WIDTH//4, 20))
        screen.blit(texto_ia, (WIDTH*3//4, 20))

        # Texto de fin de juego
        if juego_terminado:
            texto_ganador = font.render(ganador, True, GRAY)
            texto_reinicio = small_font.render("Presiona ESPACIO para reiniciar", True, GRAY)
            screen.blit(texto_ganador, (WIDTH//2 - texto_ganador.get_width()//2, HEIGHT//2 - 50))
            screen.blit(texto_reinicio, (WIDTH//2 - texto_reinicio.get_width()//2, HEIGHT//2 + 50))

        pygame.display.flip()
        clock.tick(FPS)

if __name__ == "__main__":
    main()
```
