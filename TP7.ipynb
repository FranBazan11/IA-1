1. ¿Qué representa una función de utilidad?

Una función de utilidad representa las preferencias de un agente respecto a las consecuencias o estados posibles, cuantificando cuánto se valora o importa un determinado resultado1.Puntos clave:Medida subjetiva: A diferencia del valor cuantitativo u objetivo (como el dinero), la utilidad mide la preferencia interna del agente sobre dicho resultado1.Relación no lineal: La utilidad no necesariamente crece de forma lineal con respecto a la magnitud monetaria u objetiva1.Utilidad Esperada ($EU$): Permite ponderar cada resultado por su probabilidad de ocurrencia1: $$EU(A) = \sum P(\text{resultado}_i \mid A) \cdot U(\text{resultado}_i)$$Toma de decisiones: Un agente racional selecciona la acción que maximiza su utilidad esperada1.

3. Los boletos de una lotería cuestan un dólar. Hay dos juegos posibles con distintos premios: uno de 10 dólares con una probabilidad de uno entre 50 y otro de 1.000.000 dólares con una probabilidad de uno entre 2.000.000.

&emsp;&emsp;3.1 ¿Cuál es el valor monetario esperado del boleto de lotería?

El valor esperado (VME) de cada juego es la suma de (probabilidad × premio). Como el boleto cuesta 1 dólar, conviene calcular también la ganancia neta.

Juego 1: premio de $10, probabilidad 1/50

VME = (1/50) × 10 = $0,20

Ganancia neta esperada = 0,20 − 1 = −$0,80

Juego 2: premio de $1.000.000, probabilidad 1/2.000.000

VME = (1/2.000.000) × 1.000.000 = $0,50

Ganancia neta esperada = 0,50 − 1 = −$0,50

En ambos casos el valor esperado del boleto es menor que su costo ($1), así que en promedio se pierde dinero. El segundo juego pierde menos que el primero.


&emsp;&emsp;3.2 ¿Cuándo es razonable comprar un boleto?

Si uno decide solo por el valor monetario esperado, comprar es razonable cuando VME ≥ costo del boleto, es decir, cuando la ganancia neta esperada es ≥ 0. Con estos juegos eso nunca ocurre, así que un agente neutral al riesgo no compraría ninguno.

Sin embargo, el VME no es el único criterio racional. Comprar puede ser razonable cuando:

La utilidad no es lineal en el dinero. Lo que se maximiza es la utilidad esperada, no el dinero esperado. Para alguien con muy poco dinero, $1.000.000 puede valer muchísimo más que un millón de veces la utilidad de $1 (por ejemplo, si le cambia la vida), y perder $1 casi no lo afecta. Ahí, la utilidad esperada del boleto puede superar su costo. Ese es el caso de una persona con una función de utilidad propensa al riesgo (cóncava hacia arriba) en ese rango.
El boleto tiene valor de entretenimiento. Si pagar $1 da la ilusión o la emoción de poder ganar, esa utilidad no monetaria compensa la pérdida esperada.
El premio crece. Por ejemplo, si el premio mayor se acumula hasta superar $2.000.000, el segundo juego tendría VME > $1.

### Resolución del punto 4

**Alternativas y costo esperado**

Si se repara la máquina, la avería puede ser leve, con probabilidad $1/3$ y un costo de \$300, o grave, con probabilidad $2/3$ y un costo de \$1.200. Comprar una máquina usada cuesta \$600 con certeza.

El costo esperado de reparar es:

$$
E[C_{\text{reparar}}]=\frac{1}{3}\cdot300+\frac{2}{3}\cdot1200=900.
$$

Si el criterio fuera únicamente minimizar el costo esperado, convendría **comprar**, ya que \$600 es menor que \$900. El costo esperado es un promedio: la reparación efectivamente costará \$300 o \$1.200, no \$900.

**Utilidad esperada e interpretación del gráfico**

La utilidad representa qué tan deseable es un resultado para un agente. Para calcular la utilidad esperada, primero se valora cada resultado y después se pondera por su probabilidad; en general, esto no equivale a calcular la utilidad del costo promedio.

El enunciado presenta una ambigüedad: las curvas aumentan con el dinero, pero los montos del problema son gastos. Para desarrollar una solución coherente, **suponemos que las curvas representan desutilidad del costo**, es decir, cuánto perjudica pagar cada monto. Este supuesto no está explicitado en el enunciado.

Llamamos $D(c)$ a la desutilidad de pagar un costo $c$. Como una desutilidad mayor representa un resultado peor, elegimos la alternativa con **menor desutilidad esperada**:

$$
DE(\text{reparar})=\frac{1}{3}D(300)+\frac{2}{3}D(1200),
\qquad
DE(\text{comprar})=D(600).
$$

**4.1. Agente rojo**

Leyendo aproximadamente las alturas de la curva roja:

$$
D_R(300)\approx0{,}60,\qquad D_R(600)\approx0{,}86,\qquad D_R(1200)\approx1.
$$

Por lo tanto:

$$
DE_R(\text{reparar})\approx\frac{1}{3}(0{,}60)+\frac{2}{3}(1)\approx0{,}87,
$$

$$
DE_R(\text{comprar})\approx0{,}86.
$$

Con esta lectura, comprar presenta una desutilidad ligeramente menor. Sin embargo, **la comparación queda prácticamente empatada**: la diferencia es demasiado pequeña para asegurar una preferencia a partir del dibujo. Se necesitarían valores más precisos para distinguir una leve preferencia por comprar de la indiferencia.

**4.2. Agente verde**

Leyendo aproximadamente las alturas de la curva verde:

$$
D_V(300)\approx0{,}02,\qquad D_V(600)\approx0{,}17,\qquad D_V(1200)\approx1.
$$

Entonces:

$$
DE_V(\text{reparar})\approx\frac{1}{3}(0{,}02)+\frac{2}{3}(1)\approx0{,}67,
$$

$$
DE_V(\text{comprar})\approx0{,}17.
$$

Como $0{,}17<0{,}67$, **el agente verde elige comprar la máquina usada** bajo el supuesto adoptado. En este caso, la diferencia es amplia.

Estas conclusiones dependen de interpretar las curvas como desutilidad del costo. Si representan utilidad del dinero recibido o disponible, sería necesario aclarar cómo se traducen los pagos del problema a los valores del gráfico antes de decidir.

Determine las reglas que rigen las condiciones en las que se ha de administrar una sustancia e implemente un sistema experto en CLIPS que determine si un sujeto resultará curado o no.

```python
Se calculan las entropías con las fórmulas de la Clase 8, donde p es la
cantidad de curaciones y n la de defunciones, y pi, ni son las de cada valor
del atributo:

  I(p,n)       = -p/(p+n)·log2(p/(p+n)) - n/(p+n)·log2(n/(p+n))
  E(A)         = Σ (pi+ni)/(p+n) · I(pi,ni)
  Ganancia(A)  = I(p,n) - E(A)

Las columnas numéricas (Dosis y Concentración) se categorizan con un umbral
de 75 (Poca < 75 ≤ Mucha).
En cada nodo se elige el atributo de mayor ganancia, es decir, el que más
reduce la entropía.

[Nivel 1] La entropía general es I(9,5) = 0,940. Estado tiene la mayor
ganancia (0,247; la diferencia con 0,940 - 0,694 es de redondeo), por lo que
es la raíz. Además la rama Avanzado tiene entropía 0: todos se curan, así que
no hace falta preguntar nada más.

[Nivel 2 - Incipiente] Dosis tiene ganancia 0,971, igual a la entropía del
nodo: separa los casos por completo (Poca → se cura, Mucha → muere).

[Nivel 2 - Terminal] Condición tiene ganancia 0,971 y también separa por
completo (Débil → se cura, Fuerte → muere).

La Concentración no es elegida en ningún nodo, por lo que no influye en el
resultado.

Reglas obtenidas:
1. SI Estado = Avanzado                         → se cura
2. SI Estado = Incipiente Y Dosis < 75          → se cura
3. SI Estado = Incipiente Y Dosis ≥ 75          → muere
4. SI Estado = Terminal   Y Condición = Débil   → se cura
5. SI Estado = Terminal   Y Condición = Fuerte  → muere
```

```python
"""
TP7 - Ejercicio 5: arbol de decision ID3 (entropia y ganancia, Clase 8).
Solo muestra el diagrama del arbol. Necesita matplotlib.
"""
from math import log2
import matplotlib.pyplot as plt

# ---------- Datos: (estado, concentracion, dosis, condicion, efecto) ----------
datos = [
    ("Incipiente", 75, 70, "Fuerte", "Curación"),
    ("Incipiente", 80, 90, "Fuerte", "Defunción"),
    ("Incipiente", 85, 85, "Débil",  "Defunción"),
    ("Incipiente", 62, 95, "Débil",  "Defunción"),
    ("Incipiente", 79, 70, "Débil",  "Curación"),
    ("Avanzado",   72, 90, "Fuerte", "Curación"),
    ("Avanzado",   83, 78, "Débil",  "Curación"),
    ("Avanzado",   64, 66, "Fuerte", "Curación"),
    ("Avanzado",   81, 75, "Débil",  "Curación"),
    ("Terminal",   71, 80, "Fuerte", "Defunción"),
    ("Terminal",   65, 70, "Fuerte", "Defunción"),
    ("Terminal",   75, 80, "Débil",  "Curación"),
    ("Terminal",   68, 80, "Débil",  "Curación"),
    ("Terminal",   70, 96, "Débil",  "Curación"),
]
UMBRAL = 75  # Poca < 75 <= Mucha

def categorizar(f):
    est, conc, dosis, cond, efecto = f
    return {
        "Estado": est,
        "Concentración": "Mucha" if conc >= UMBRAL else "Poca",
        "Dosis": "Mucha" if dosis >= UMBRAL else "Poca",
        "Condición": cond,
        "Efecto": efecto,
    }

ejemplos = [categorizar(f) for f in datos]
ATRIBUTOS = ["Estado", "Concentración", "Dosis", "Condición"]

# ---------- Entropía y ganancia ----------
def conteo(filas):
    p = sum(f["Efecto"] == "Curación" for f in filas)
    return p, len(filas) - p

def entropia(filas):
    """I(p,n) = -p/(p+n)*log2(p/(p+n)) - n/(p+n)*log2(n/(p+n))"""
    total = len(filas)
    if total == 0:
        return 0.0
    r = 0.0
    for x in conteo(filas):
        if x:
            r -= (x / total) * log2(x / total)
    return r

def entropia_atributo(filas, atr):
    """E(A) = suma de (pi+ni)/(p+n) * I(pi,ni)"""
    e = 0.0
    for v in {f[atr] for f in filas}:
        sub = [f for f in filas if f[atr] == v]
        e += len(sub) / len(filas) * entropia(sub)
    return e

def ganancia(filas, atr):
    return entropia(filas) - entropia_atributo(filas, atr)

# ---------- ID3 ----------
def id3(filas, atributos):
    """Hoja (str) o nodo (atributo, ganancia, I_nodo, ramas).
    ramas = {valor: (subarbol, I_rama, p_rama, n_rama)}"""
    efectos = {f["Efecto"] for f in filas}
    if len(efectos) == 1:                      # nodo puro -> hoja
        return efectos.pop()
    if not atributos:                          # sin atributos -> mayoria
        return max(efectos, key=lambda e: sum(f["Efecto"] == e for f in filas))
    mejor = max(atributos, key=lambda a: ganancia(filas, a))
    ramas = {}
    for v in sorted({f[mejor] for f in filas}):
        sub = [f for f in filas if f[mejor] == v]
        p, n = conteo(sub)
        ramas[v] = (id3(sub, [a for a in atributos if a != mejor]),
                    entropia(sub), p, n)
    return (mejor, ganancia(filas, mejor), entropia(filas), ramas)

# ---------- Dibujo ----------
def contar_hojas(nodo):
    if not isinstance(nodo, tuple):
        return 1
    return sum(contar_hojas(r[0]) for r in nodo[3].values())

def dibujar(nodo, ax, x0, x1, y, dy=1.3, padre=None, etiqueta=""):
    x = (x0 + x1) / 2
    if padre:
        ax.plot([padre[0], x], [padre[1] - 0.2, y + 0.2], "k-", lw=1)
        ax.text((padre[0] + x) / 2, (padre[1] + y) / 2, etiqueta,
                ha="center", va="center", fontsize=9, color="navy",
                bbox=dict(fc="white", ec="navy", boxstyle="round,pad=0.25"))
    if not isinstance(nodo, tuple):            # hoja
        color = "#9be28d" if nodo == "Curación" else "#f19a9a"
        ax.text(x, y, nodo, ha="center", va="center", fontsize=11,
                bbox=dict(boxstyle="round", fc=color, ec="k"))
        return
    atr, g, i_nodo, ramas = nodo
    ax.text(x, y, f"{atr}\nI = {i_nodo:.3f}\nGanancia = {g:.3f}", ha="center",
            va="center", fontsize=10,
            bbox=dict(boxstyle="round", fc="#cfe2ff", ec="k"))
    total = contar_hojas(nodo)
    cur = x0
    for valor, (hijo, i_r, p, n) in ramas.items():
        ancho = (x1 - x0) * contar_hojas(hijo) / total
        et = f"{valor}\nI = {i_r:.3f}\n({p} cur, {n} def)"
        dibujar(hijo, ax, cur, cur + ancho, y - dy, dy, (x, y), et)
        cur += ancho

if __name__ == "__main__":
    arbol = id3(ejemplos, ATRIBUTOS)
    fig, ax = plt.subplots(figsize=(13, 8))
    dibujar(arbol, ax, 0, 10, 3.9)
    ax.set_xlim(0, 10); ax.set_ylim(0.9, 4.5); ax.axis("off")
    ax.set_title("Árbol de decisión ID3 - Ejercicio 5 (TP7)\n"
                 "(I = entropía de cada rama; cur = curaciones, def = defunciones)")
    plt.tight_layout()
    plt.show()
```

## Codigo en CLIPS

;;; TP7 - Ejercicio 5: sistema experto (curacion / defuncion)

;;; ---- Pregunta 1 (siempre): la de mayor ganancia ----
(defrule leer-estado
	=>
	(printout t "Estado de la enfermedad (incipiente/avanzado/terminal): ")
	(assert (estado (read)))
)

;;; ---- Rama AVANZADO: entropia 0, no se pregunta nada mas ----
(defrule estado-avanzado
	(estado ?e)
	=>
	(if (eq ?e avanzado)
	then
		(assert (efecto curacion))
	)
)

;;; ---- Rama INCIPIENTE: se pregunta solo la dosis (umbral 75) ----
(defrule leer-dosis
	(estado ?e)
	=>
	(if (eq ?e incipiente)
	then
		(printout t "Numero de dosis: ")
		(assert (dosis (read)))
	)
)

(defrule incipiente-dosis
	(estado ?e)
	(dosis ?d)
	=>
	(if (eq ?e incipiente)
	then
		(if (< ?d 75)
		then
			(assert (efecto curacion))
		else
			(assert (efecto defuncion))
		)
	)
)

;;; ---- Rama TERMINAL: se pregunta solo la condicion fisica ----
(defrule leer-condicion
	(estado ?e)
	=>
	(if (eq ?e terminal)
	then
		(printout t "Condicion fisica (fuerte/debil): ")
		(assert (condicion (read)))
	)
)

(defrule terminal-condicion
	(estado ?e)
	(condicion ?c)
	=>
	(if (eq ?e terminal)
	then
		(if (eq ?c debil)
		then
			(assert (efecto curacion))
		else
			(assert (efecto defuncion))
		)
	)
)

;;; ---- Resultado ----
(defrule resultado
	(efecto ?x)
	(estado ?)
	=>
	(if (eq ?x curacion)
	then

- Una pregunta aporta mucha o poca información según cuánto reduce la

- Fórmulas: I(p,n) = -p/(p+n)·log2(p/(p+n)) - n/(p+n)·log2(n/(p+n)),

- CLIPS: los facts son los hechos sensados; las reglas definen el orden de
