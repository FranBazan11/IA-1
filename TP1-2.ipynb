# Ejercicios Teóricos

## 1. Defina con sus propias palabras inteligencia natural, inteligencia artificial y agente.

Inteligencia natural: es la capacidad que tenemos los seres vivos, especialmente las personas, para aprender, razonar, resolver problemas, adaptarnos a situaciones nuevas y tomar decisiones a partir de nuestras experiencias.
Inteligencia artificial: es la capacidad que se les da a las máquinas o programas para realizar tareas que normalmente requieren inteligencia humana, como aprender de datos, reconocer imágenes, comprender lenguaje o tomar decisiones.
Agente: es un sistema que percibe lo que ocurre en su entorno, procesa esa información y realiza acciones para alcanzar un objetivo. Por ejemplo, un robot que detecta un obstáculo y cambia su dirección.

## 2. ¿Qué es un agente racional?

Un agente racional es un agente que elige la acción que considera más conveniente para alcanzar su objetivo, teniendo en cuenta la información que tiene disponible y las posibles consecuencias de sus acciones.

## 3. ¿Un agente es siempre una computadora?

**No, un agente no es necesariamente una computadora.**

Desde una perspectiva de ingeniería mecatrónica, un agente es una **entidad autónoma capaz de percibir su entorno y tomar acciones racionalmente para alcanzar objetivos específicos**. Esta definición es mucho más amplia que el concepto de "computadora".

Ejemplos de agentes no-computacionales:

1. **Sistemas Mecánicos**: Un termostato analógico es un agente que percibe temperatura y actúa regulando la calefacción sin procesamiento digital.

2. **Organismos Biológicos**: Una bacteria que se mueve hacia fuentes de nutrientes es un agente que percibe y actúa autónomamente.

3. **Sistemas Híbridos**: Un robot móvil con controladores mecánicos y sensores actúa como agente, aunque combine componentes eléctricos y mecánicos.

En ingeniería mecatrónica:

Los agentes pueden ser:
- **Puramente mecánicos**: Sistemas de control basados en mecanismos físicos
- **Electromecánicos**: Combinación de circuitos electrónicos y componentes mecánicos
- **Computacionales**: Sistemas con microcontroladores y procesadores digitales

**Conclusión**: La computadora es solo una **implementación posible** de un agente. Lo que define a un agente es su capacidad de autonomía, percepción y acción racional, independientemente de su naturaleza física o componentes internos.

## 4. Defina Omnisciencia, Aprendizaje y Autonomía.

### 1. Omnisciencia

**Definición**: Es la capacidad de un agente de tener **conocimiento completo y perfecto de todo lo que ocurre en su entorno**, incluyendo todas las consecuencias futuras de sus acciones, sin incertidumbre.

**Perspectiva Mecatrónica**:
- En sistemas de control clásicos (PLC, controladores industriales), la omnisciencia es **inalcanzable** porque:
  - Los sensores tienen limitaciones físicas y ruido
  - Existen retrasos en la transmisión de datos
  - El entorno es dinámico y parcialmente observable
  
- **Ejemplo**: Un robot industrial que posiciona piezas jamás conoce con exactitud la posición real (hay tolerancias, fricción, desviaciones).

**Implicación práctica**: Los agentes reales trabajan bajo **incertidumbre**, desarrollando estrategias robustas a través de retroalimentación y adaptación.

---

### 2. Aprendizaje

**Definición**: Es la capacidad de un agente de **mejorar su desempeño a través de la experiencia**, modificando su comportamiento o modelo interno del entorno sin ser explícitamente programado para cada situación.

**Perspectiva Mecatrónica**:
- **Aprendizaje por control adaptativo**: Los sistemas ajustan parámetros automáticamente (ej: PID adaptativos)
- **Aprendizaje por refuerzo**: El agente recibe retroalimentación (recompensas/castigos) y optimiza sus acciones
- **Aprendizaje supervisado**: Entrenar redes neuronales en microcontroladores para reconocer patrones de sensores

**Ejemplos**:
- Robot que aprende a caminar ajustando actuadores basándose en caídas previas
- Brazo robótico que calibra su precisión después de múltiples intentos
- Sistema de visión que mejora la detección de objetos con más imágenes de entrenamiento

**Ventaja**: Permite al agente adaptarse a cambios no previstos sin reprogramación.

---

### 3. Autonomía

**Definición**: Es la capacidad de un agente de **tomar decisiones y actuar de forma independiente**, sin intervención externa constante, basándose en sus propios objetivos y percepciones del entorno.

**Perspectiva Mecatrónica**:
- **Autonomía parcial**: El agente opera dentro de límites predefinidos (ej: robot vacuum con zonas prohibidas)
- **Autonomía total**: Capacidad de autogestión completa, estableciendo y modificando sus propios objetivos
- **Niveles de autonomía**:
  1. **Teleoperado**: Controlado completamente por operador humano
  2. **Semi-autónomo**: Requiere supervisión pero toma decisiones locales
  3. **Autónomo**: Opera independientemente con mínima intervención

**Ejemplos**:
- Vehículo autónomo que navega sin intervención pero respeta reglas de tráfico predefinidas
- Drone explorador que planifica rutas automáticamente dentro de su misión
- Robot de servicio que gestiona autónomamente su carga de batería sin comandos externos

**Limitación importante**: La autonomía real siempre está **limitada por restricciones de diseño, seguridad y recursos**.

---

### Interrelación en Sistemas Mecatrónicos

| Aspecto | Impacto |
|--------|--------|
| **Sin omnisciencia** | El agente debe usar sensores y lógica probabilística |
| **Con aprendizaje** | Se adapta a incertidumbre y cambios ambientales |
| **Con autonomía** | Ejecuta decisiones sin depender de operadores humanos |

**Conclusión**: Los sistemas mecatrónicos inteligentes balancean estos tres elementos para crear agentes efectivos bajo incertidumbre, capaces de mejorar continuamente y actuar independientemente dentro de sus capacidades.

## 5. Defina cada tipo de agente en función de su **estructura** y dé un ejemplo de cada categoría.

Existen cinco tipos principales de agentes según la complejidad de su programa y estructura:

Agente reactivo simple: Su estructura se basa en reglas de condición-acción. Actúan únicamente según la percepción actual, sin considerar el historial de percepciones
. Requieren generalmente un entorno totalmente observable para tomar la decisión correcta
.
Ejemplo: Una aspiradora sin sensor de ubicación que simplemente aplica la regla: "si el cuadro actual está sucio, aspirar"

.
Agente reactivo basado en modelos: A diferencia del anterior, mantiene un estado interno que le permite seguir el rastro de aspectos del mundo que no son evidentes en la percepción actual
. Su estructura incorpora conocimiento sobre cómo evoluciona el entorno y cuáles son los efectos de sus propias acciones
.
Ejemplo: Un termostato inteligente que utiliza una ecuación para modelar la inercia térmica de una habitación, permitiéndole predecir cambios de temperatura y actuar de forma más eficiente

.
Agente basado en objetivos: Además del estado interno y el modelo del mundo, este agente cuenta con información sobre sus objetivos
. Su estructura le permite realizar procesos de búsqueda y planificación, proyectando qué sucederá si realiza una acción determinada para verificar si esta lo acerca a su meta
.
Ejemplo: Un agente de navegación por GPS para entrega de paquetes que analiza un grafo de calles para encontrar la ruta que garantice la entrega en el menor tiempo posible

.
Agente basado en utilidad: Utiliza una función de utilidad para elegir la mejor acción entre diversas alternativas que podrían llevar al objetivo
. Su estructura permite cuantificar el nivel de "felicidad" o éxito de un estado, lo que resulta fundamental cuando hay objetivos en conflicto o se debe equilibrar el rendimiento
.
Ejemplo: Un sistema de control de climatización en un edificio que debe decidir entre maximizar el confort térmico de los empleados o minimizar el costo de energía eléctrica

.
Agente que aprende: Su estructura se descompone en cuatro componentes conceptuales: el elemento de aprendizaje (realiza mejoras), el elemento de actuación (selecciona acciones externas), la crítica (proporciona realimentación basada en un estándar de actuación) y el generador de problemas (propone acciones exploratorias)
. Este proceso permite al agente mejorar su desempeño futuro basándose en su experiencia previa
.
Ejemplo: Un agente para jugar al Ti-Te-Ti (tres en línea) que, tras cada partida, recibe realimentación sobre si ganó o perdió y actualiza su conocimiento interno para refinar su estrategia en el siguiente juego

### Resolución del ejercicio 8

### Tipo de agente

La Hormiga de Langton es un **agente reactivo simple**: elige qué hacer a partir del color de la casilla actual. No planifica, no aprende y no necesita conocer las casillas anteriores para aplicar su regla. Su orientación forma parte de su estado actual y determina hacia dónde avanzará luego de girar.

### Tabla REAS

| Elemento | Descripción |
|---|---|
| **Rendimiento** | Aplicar correctamente la regla en cada iteración, actualizar la casilla y avanzar sin perder su posición u orientación. |
| **Entorno** | Una cuadrícula bidimensional infinita cuyas casillas pueden ser blancas o negras. |
| **Actuadores** | Cambiar el color de la casilla, girar 90° a la derecha o izquierda y avanzar una casilla. |
| **Sensores** | Percibir el color de la casilla actual y conocer su posición y orientación. |

### Propiedades del entorno

- **Parcialmente observable:** la hormiga solo percibe la casilla en la que se encuentra, no el estado completo del tablero. Esa percepción local, sin embargo, es suficiente para decidir su acción inmediata.
- **Determinista:** dado el tablero, la posición y la orientación actuales, el estado siguiente queda definido de manera única. La elección aleatoria inicial no vuelve aleatorias las transiciones posteriores.
- **Secuencial:** cada cambio de color modifica el tablero y puede afectar decisiones futuras cuando la hormiga regrese a esa casilla.
- **Estático:** el tablero no cambia mientras la hormiga decide; solo cambia como consecuencia de sus acciones.
- **Discreto:** las casillas, los colores, las orientaciones y los movimientos pertenecen a conjuntos finitos o numerables.
- **Conocido:** las reglas y los efectos de cada acción están definidos de antemano.
- **Agente individual:** no hay otros agentes que modifiquen el entorno.

### Reglas aplicadas en cada caso

| Casilla percibida | Nuevo color | Giro | Movimiento |
|---|---|---|---|
| Blanca | Negra | 90° a la derecha | Avanza una casilla |
| Negra | Blanca | 90° a la izquierda | Avanza una casilla |

El tablero se representa con un conjunto que contiene solamente las coordenadas de las casillas negras. Una coordenada ausente se interpreta como blanca. De esta manera se simula una cuadrícula sin bordes y se evita reservar memoria para casillas que nunca fueron visitadas.

```python
from __future__ import annotations

import random
from collections.abc import Mapping, Sequence

import matplotlib.pyplot as plt


Posicion = tuple[int, int]
Captura = tuple[frozenset[Posicion], Posicion, int]

# El orden permite girar sumando o restando uno al índice actual.
DIRECCIONES: tuple[Posicion, ...] = (
    (-1, 0),  # Norte
    (0, 1),   # Este
    (1, 0),   # Sur
    (0, -1),  # Oeste
)


def ejecutar_paso(
    posicion: Posicion,
    direccion: int,
    casillas_negras: set[Posicion],
) -> tuple[Posicion, int, str]:
    """Aplica una vez las reglas y devuelve la nueva pose de la hormiga."""
    if posicion in casillas_negras:
        casillas_negras.remove(posicion)
        direccion = (direccion - 1) % len(DIRECCIONES)
        giro = "I"
    else:
        casillas_negras.add(posicion)
        direccion = (direccion + 1) % len(DIRECCIONES)
        giro = "D"

    fila, columna = posicion
    cambio_fila, cambio_columna = DIRECCIONES[direccion]
    nueva_posicion = (fila + cambio_fila, columna + cambio_columna)
    return nueva_posicion, direccion, giro


def simular_hormiga(
    iteraciones: int,
    semilla: int = 42,
    capturar_en: Sequence[int] = (),
    radio_inicio: int = 10,
) -> tuple[list[str], list[Posicion], dict[int, Captura]]:
    """Simula la hormiga sobre un tablero inicialmente blanco e infinito."""
    if iteraciones < 0:
        raise ValueError("La cantidad de iteraciones no puede ser negativa.")
    if radio_inicio < 0:
        raise ValueError("El radio inicial no puede ser negativo.")
    if any(paso < 0 or paso > iteraciones for paso in capturar_en):
        raise ValueError("Cada captura debe pertenecer al rango de la simulación.")

    generador = random.Random(semilla)
    posicion = (
        generador.randint(-radio_inicio, radio_inicio),
        generador.randint(-radio_inicio, radio_inicio),
    )
    direccion = generador.randrange(len(DIRECCIONES))
    casillas_negras: set[Posicion] = set()
    giros: list[str] = []
    posiciones = [posicion]
    capturas: dict[int, Captura] = {}
    pasos_a_capturar = set(capturar_en)

    if 0 in pasos_a_capturar:
        capturas[0] = (frozenset(), posicion, direccion)

    for paso in range(1, iteraciones + 1):
        posicion, direccion, giro = ejecutar_paso(
            posicion, direccion, casillas_negras
        )
        giros.append(giro)
        posiciones.append(posicion)

        if paso in pasos_a_capturar:
            capturas[paso] = (
                frozenset(casillas_negras),
                posicion,
                direccion,
            )

    return giros, posiciones, capturas


def detectar_inicio_periodo(
    giros: Sequence[str],
    periodo: int = 104,
    ciclos_minimos: int = 10,
) -> int | None:
    """Busca el primer índice desde el cual los giros repiten el período dado."""
    if periodo <= 0:
        raise ValueError("El período debe ser positivo.")
    if ciclos_minimos < 2:
        raise ValueError("Se necesitan al menos dos ciclos para comparar.")

    ultimo_inicio = len(giros) - periodo * ciclos_minimos
    for inicio in range(ultimo_inicio + 1):
        se_repite = all(
            giros[indice] == giros[indice + periodo]
            for indice in range(inicio, len(giros) - periodo)
        )
        if se_repite:
            return inicio

    return None


def graficar_capturas(capturas: Mapping[int, Captura]) -> None:
    """Grafica el tablero y la pose de la hormiga en cada captura."""
    if not capturas:
        raise ValueError("Debe proporcionarse al menos una captura.")

    cantidad = len(capturas)
    filas = (cantidad + 1) // 2
    figura, ejes = plt.subplots(
        filas, 2, figsize=(12, 5 * filas), squeeze=False, constrained_layout=True
    )
    nombres_etapas = {
        100: "estructura inicial",
        1_000: "fase caótica",
        9_977: "fin del transitorio",
        12_000: "autopista",
    }

    for eje, (paso, captura) in zip(ejes.flat, sorted(capturas.items())):
        casillas_negras, posicion, direccion = captura
        if casillas_negras:
            coordenadas = list(casillas_negras)
            filas_negras, columnas_negras = zip(*coordenadas)
            eje.scatter(
                columnas_negras,
                [-fila for fila in filas_negras],
                c="black",
                marker="s",
                s=12,
            )
            amplitud = max(
                max(columnas_negras) - min(columnas_negras),
                max(filas_negras) - min(filas_negras),
                1,
            )
        else:
            amplitud = 1

        fila, columna = posicion
        cambio_fila, cambio_columna = DIRECCIONES[direccion]
        longitud_flecha = max(1.5, amplitud * 0.06)
        eje.scatter(columna, -fila, c="crimson", s=45, zorder=3)
        eje.annotate(
            "",
            xy=(
                columna + longitud_flecha * cambio_columna,
                -fila - longitud_flecha * cambio_fila,
            ),
            xytext=(columna, -fila),
            arrowprops={"arrowstyle": "->", "color": "crimson", "lw": 2},
        )
        etapa = nombres_etapas.get(paso, "captura")
        eje.set_title(f"{paso:,} iteraciones — {etapa}".replace(",", "."))
        eje.set_xlabel("Columna")
        eje.set_ylabel("Fila")
        eje.set_aspect("equal")
        eje.grid(alpha=0.15)
        eje.margins(0.08)

    for eje in ejes.flat[cantidad:]:
        eje.remove()

    figura.suptitle("Evolución de la Hormiga de Langton", fontsize=16)
    plt.show()
```

### Ejecución y comprobaciones

Se simulan 12.000 iteraciones. La semilla solo hace reproducibles la posición y la orientación iniciales; como el tablero infinito comienza completamente blanco, cambiar esos valores únicamente traslada o rota el mismo comportamiento. El detector compara la secuencia de giros y exige observar al menos diez ciclos completos antes de aceptar una periodicidad.

```python
ITERACIONES = 12_000
PERIODO_AUTOPISTA = 104
PASOS_CAPTURA = (100, 1_000, 9_977, 12_000)

# Comprobación independiente de las dos reglas del enunciado.
tablero_blanco: set[Posicion] = set()
posicion_blanco, direccion_blanco, giro_blanco = ejecutar_paso(
    (0, 0), 0, tablero_blanco
)
assert tablero_blanco == {(0, 0)}
assert (posicion_blanco, direccion_blanco, giro_blanco) == ((0, 1), 1, "D")

tablero_negro: set[Posicion] = {(0, 0)}
posicion_negro, direccion_negro, giro_negro = ejecutar_paso(
    (0, 0), 0, tablero_negro
)
assert tablero_negro == set()
assert (posicion_negro, direccion_negro, giro_negro) == ((0, -1), 3, "I")

giros, posiciones, capturas = simular_hormiga(
    ITERACIONES, semilla=42, capturar_en=PASOS_CAPTURA
)
inicio_periodo = detectar_inicio_periodo(giros, PERIODO_AUTOPISTA)
if inicio_periodo is None:
    raise RuntimeError("No se detectó la autopista en las iteraciones simuladas.")

# La misma semilla debe producir exactamente el mismo recorrido.
assert simular_hormiga(200, semilla=42) == simular_hormiga(200, semilla=42)
assert inicio_periodo == 9_977
assert giros[inicio_periodo:-PERIODO_AUTOPISTA] == giros[
    inicio_periodo + PERIODO_AUTOPISTA :
]

posicion_inicial_ciclo = posiciones[inicio_periodo]
posicion_final_ciclo = posiciones[inicio_periodo + PERIODO_AUTOPISTA]
desplazamiento = tuple(
    final - inicial
    for inicial, final in zip(posicion_inicial_ciclo, posicion_final_ciclo)
)
assert all(abs(cambio) == 2 for cambio in desplazamiento)

inicio_formateado = f"{inicio_periodo:,}".replace(",", ".")
movimiento_formateado = f"{inicio_periodo + 1:,}".replace(",", ".")
print("Comprobaciones de las reglas: correctas.")
print(
    f"Inicio de la autopista: después de {inicio_formateado} iteraciones "
    f"completadas (movimiento {movimiento_formateado})."
)
print(f"Período detectado: {PERIODO_AUTOPISTA} movimientos.")
print(f"Desplazamiento por período (fila, columna): {desplazamiento}.")
```

```python
graficar_capturas(capturas)
```

### ¿Se repite algún patrón?

Sí. En las capturas se distinguen tres comportamientos:

1. **Estructura inicial:** durante los primeros cientos de movimientos aparecen figuras relativamente compactas y simétricas.
2. **Fase caótica:** la trayectoria pierde esa regularidad visible y la hormiga recorre repetidamente una región limitada.
3. **Orden emergente:** después de completar **9.977 iteraciones**, la secuencia que comienza con el movimiento **9.978** se repite cada **104 movimientos**. En cada ciclo la hormiga se desplaza dos casillas en ambos ejes, por lo que construye una estructura diagonal conocida como *autopista*.

La diferencia entre 9.977 y 9.978 se debe únicamente a cómo se numeran los estados: el índice 9.977 de la lista de giros corresponde al movimiento 9.978 cuando se empieza a contar en uno. La detección realizada aquí es experimental sobre las 12.000 iteraciones simuladas.

## Ejercicio 9

9. El Juego de la Vida de Conway consiste en un tablero donde cada casilla representa una célula, de manera que a cada célula le rodean 8 vecinas. Las células tienen dos estados: están *vivas* o *muertas*. En cada iteración, el estado de todas las células se tiene en cuenta para calcular el estado siguiente en simultáneo de acuerdo a las siguientes acciones:

* Nacer: Si una célula muerta tiene exactamente 3 células vecinas vivas, dicha célula pasa a estar viva.

* Morir: Una célula viva puede morir sobrepoblación cuando tiene más de tres vecinos alrededor o por aislamiento si tiene solo un vecino o ninguno.

* Vivir: una célula se mantiene viva si tiene 2 o 3 vecinos a su alrededor.

    Caracterice el agente con su tabla REAS y las propiedades del entorno para después programarlo en Python:

```python
from collections import Counter
from collections.abc import Iterable, Mapping

import matplotlib.pyplot as plt
from IPython.display import Markdown, display


display(Markdown("""
## Resolución del ejercicio 9

### Tipo de agente

Cada célula es un **agente reactivo simple**: percibe su estado y el de sus ocho
vecinas, y aplica una regla condición-acción. No posee memoria, aprendizaje ni
planificación. Las decisiones se calculan sobre el mismo estado del tablero y se
aplican simultáneamente.

### Tabla REAS

| Elemento | Descripción |
|---|---|
| **Rendimiento** | Aplicar correctamente nacimiento, supervivencia y muerte, actualizando todas las células en simultáneo. |
| **Entorno** | Una cuadrícula bidimensional de células vivas o muertas, cada una con ocho vecinas. |
| **Actuadores** | Establecer su estado para la siguiente generación: viva o muerta. |
| **Sensores** | Percibir su estado actual y contar sus vecinas vivas. |

### Propiedades del entorno

- **Parcialmente observable:** cada célula percibe solamente su vecindad.
- **Determinista:** cada generación determina de forma única la siguiente.
- **Secuencial:** una generación condiciona todas las posteriores.
- **Dinámico:** la vecindad cambia entre generaciones por la acción de las demás células.
- **Discreto:** posiciones, estados y generaciones son discretos.
- **Conocido:** las reglas y sus efectos están definidos de antemano.
- **Multiagente:** las células actúan simultáneamente e influyen en sus vecinas.
"""))


Celula = tuple[int, int]
Estado = frozenset[Celula]
VECINAS: tuple[Celula, ...] = tuple(
    (df, dc)
    for df in (-1, 0, 1)
    for dc in (-1, 0, 1)
    if (df, dc) != (0, 0)
)


def siguiente_generacion(celulas_vivas: Iterable[Celula]) -> set[Celula]:
    """Calcula simultáneamente la siguiente generación."""
    vivas = set(celulas_vivas)
    cantidad_vecinas: Counter[Celula] = Counter()

    for fila, columna in vivas:
        for df, dc in VECINAS:
            cantidad_vecinas[(fila + df, columna + dc)] += 1

    return {
        celula
        for celula, cantidad in cantidad_vecinas.items()
        if cantidad == 3 or (cantidad == 2 and celula in vivas)
    }


def simular_vida(
    celulas_iniciales: Iterable[Celula], generaciones: int
) -> dict[int, Estado]:
    """Simula y devuelve todos los estados desde 0 hasta generaciones."""
    if generaciones < 0:
        raise ValueError("La cantidad de generaciones no puede ser negativa.")

    vivas = set(celulas_iniciales)
    estados = {0: frozenset(vivas)}
    for generacion in range(1, generaciones + 1):
        vivas = siguiente_generacion(vivas)
        estados[generacion] = frozenset(vivas)
    return estados


def graficar_generaciones(estados: Mapping[int, Estado]) -> None:
    """Grafica los estados recibidos usando los mismos límites."""
    if not estados:
        raise ValueError("Debe proporcionarse al menos una generación.")

    ordenados = sorted(estados.items())
    todas = set().union(*(estado for _, estado in ordenados))
    if todas:
        filas, columnas = zip(*todas)
        limites_x = (min(columnas) - 1, max(columnas) + 1)
        limites_y = (-max(filas) - 1, -min(filas) + 1)
    else:
        limites_x = limites_y = (-1, 1)

    figura, ejes = plt.subplots(
        1, len(ordenados), figsize=(3.2 * len(ordenados), 3.5),
        squeeze=False, constrained_layout=True
    )
    for eje, (generacion, vivas) in zip(ejes.flat, ordenados):
        if vivas:
            filas, columnas = zip(*vivas)
            eje.scatter(
                columnas, [-fila for fila in filas],
                color="black", marker="s", s=180
            )
        eje.set_title(f"Generación {generacion}")
        eje.set_xlim(*limites_x)
        eje.set_ylim(*limites_y)
        eje.set_aspect("equal")
        eje.grid(alpha=0.25)
        eje.tick_params(labelbottom=False, labelleft=False)

    figura.suptitle("Juego de la Vida: evolución de un planeador", fontsize=15)
    plt.show()


# Comprobación de las reglas del enunciado.
entorno_nacimiento = {(-1, -1), (-1, 0), (-1, 1)}
assert (0, 0) in siguiente_generacion(entorno_nacimiento)

parpadeador_horizontal = {(0, -1), (0, 0), (0, 1)}
parpadeador_vertical = {(-1, 0), (0, 0), (1, 0)}
assert (0, 0) in siguiente_generacion(parpadeador_horizontal)  # Vive con 2.
entorno_supervivencia_tres = {(0, 0), (-1, -1), (-1, 0), (-1, 1)}
assert (0, 0) in siguiente_generacion(entorno_supervivencia_tres)  # Vive con 3.

assert (0, 0) not in siguiente_generacion({(0, 0)})  # Muere con 0.
assert (0, 0) not in siguiente_generacion({(0, 0), (0, 1)})  # Muere con 1.
entorno_sobrepoblado = {(0, 0), (-1, 0), (1, 0), (0, -1), (0, 1)}
assert (0, 0) not in siguiente_generacion(entorno_sobrepoblado)  # Muere con 4.

# El parpadeador verifica que la actualización sea simultánea.
assert siguiente_generacion(parpadeador_horizontal) == parpadeador_vertical
assert siguiente_generacion(parpadeador_vertical) == parpadeador_horizontal

# Un planeador recupera su forma, desplazada (1, 1), cada 4 generaciones.
planeador = {(0, 1), (1, 2), (2, 0), (2, 1), (2, 2)}
generaciones_planeador = simular_vida(planeador, 4)
planeador_desplazado = {(fila + 1, columna + 1) for fila, columna in planeador}
assert generaciones_planeador[4] == planeador_desplazado

print("Comprobaciones de las reglas: correctas.")
print("Parpadeador: período de 2 generaciones.")
print("Planeador: desplazamiento (1, 1) cada 4 generaciones.")
graficar_generaciones(generaciones_planeador)
```
