1. Defina con sus propias palabras inteligencia natural, inteligencia artificial y agente.

Inteligencia natural: es la capacidad que tenemos los seres vivos, especialmente las personas, para aprender, razonar, resolver problemas, adaptarnos a situaciones nuevas y tomar decisiones a partir de nuestras experiencias.
Inteligencia artificial: es la capacidad que se les da a las máquinas o programas para realizar tareas que normalmente requieren inteligencia humana, como aprender de datos, reconocer imágenes, comprender lenguaje o tomar decisiones.
Agente: es un sistema que percibe lo que ocurre en su entorno, procesa esa información y realiza acciones para alcanzar un objetivo. Por ejemplo, un robot que detecta un obstáculo y cambia su dirección.


2. ¿Qué es un agente racional?

Un agente racional es un agente que elige la acción que considera más conveniente para alcanzar su objetivo, teniendo en cuenta la información que tiene disponible y las posibles consecuencias de sus acciones.

3. ¿Un agente es siempre una computadora?

**No, un agente no es necesariamente una computadora.** 

Desde una perspectiva de ingeniería mecatrónica, un agente es una **entidad autónoma capaz de percibir su entorno y tomar acciones racionalmente para alcanzar objetivos específicos**. Esta definición es mucho más amplia que el concepto de "computadora".

### Ejemplos de agentes no-computacionales:

1. **Sistemas Mecánicos**: Un termostato analógico es un agente que percibe temperatura y actúa regulando la calefacción sin procesamiento digital.

2. **Organismos Biológicos**: Una bacteria que se mueve hacia fuentes de nutrientes es un agente que percibe y actúa autónomamente.

3. **Sistemas Híbridos**: Un robot móvil con controladores mecánicos y sensores actúa como agente, aunque combine componentes eléctricos y mecánicos.

### En ingeniería mecatrónica:

Los agentes pueden ser:
- **Puramente mecánicos**: Sistemas de control basados en mecanismos físicos
- **Electromecánicos**: Combinación de circuitos electrónicos y componentes mecánicos
- **Computacionales**: Sistemas con microcontroladores y procesadores digitales

**Conclusión**: La computadora es solo una **implementación posible** de un agente. Lo que define a un agente es su capacidad de autonomía, percepción y acción racional, independientemente de su naturaleza física o componentes internos.

## 1. Omnisciencia

**Definición**: Es la capacidad de un agente de tener **conocimiento completo y perfecto de todo lo que ocurre en su entorno**, incluyendo todas las consecuencias futuras de sus acciones, sin incertidumbre.

**Perspectiva Mecatrónica**:
- En sistemas de control clásicos (PLC, controladores industriales), la omnisciencia es **inalcanzable** porque:
  - Los sensores tienen limitaciones físicas y ruido
  - Existen retrasos en la transmisión de datos
  - El entorno es dinámico y parcialmente observable
  
- **Ejemplo**: Un robot industrial que posiciona piezas jamás conoce con exactitud la posición real (hay tolerancias, fricción, desviaciones).

**Implicación práctica**: Los agentes reales trabajan bajo **incertidumbre**, desarrollando estrategias robustas a través de retroalimentación y adaptación.

---

## 2. Aprendizaje

**Definición**: Es la capacidad de un agente de **mejorar su desempeño a través de la experiencia**, modificando su comportamiento o modelo interno del entorno sin ser explícitamente programado para cada situación.

**Perspectiva Mecatrónica**:
- **Aprendizaje por control adaptativo**: Los sistemas ajustan parámetros automáticamente (ej: PID adaptativos)
- **Aprendizaje por refuerzo**: El agente recibe retroalimentación (recompensas/castigos) y optimiza sus acciones
- **Aprendizaje supervisado**: Entrenar redes neuronales en microcontroladores para reconocer patrones de sensores

**Ejemplos**:
- Robot que aprende a caminar ajustando actuadores basándose en caídas previas
- Brazo robótico que calibra su precisión después de múltiples intentos
- Sistema de visión que mejora la detección de objetos con más imágenes de entrenamiento

**Ventaja**: Permite al agente adaptarse a cambios no previstos sin reprogramación.

---

## 3. Autonomía

**Definición**: Es la capacidad de un agente de **tomar decisiones y actuar de forma independiente**, sin intervención externa constante, basándose en sus propios objetivos y percepciones del entorno.

**Perspectiva Mecatrónica**:
- **Autonomía parcial**: El agente opera dentro de límites predefinidos (ej: robot vacuum con zonas prohibidas)
- **Autonomía total**: Capacidad de autogestión completa, estableciendo y modificando sus propios objetivos
- **Niveles de autonomía**:
  1. **Teleoperado**: Controlado completamente por operador humano
  2. **Semi-autónomo**: Requiere supervisión pero toma decisiones locales
  3. **Autónomo**: Opera independientemente con mínima intervención

**Ejemplos**:
- Vehículo autónomo que navega sin intervención pero respeta reglas de tráfico predefinidas
- Drone explorador que planifica rutas automáticamente dentro de su misión
- Robot de servicio que gestiona autónomamente su carga de batería sin comandos externos

**Limitación importante**: La autonomía real siempre está **limitada por restricciones de diseño, seguridad y recursos**.

## Interrelación en Sistemas Mecatrónicos

| Aspecto | Impacto |
|--------|--------|
| **Sin omnisciencia** | El agente debe usar sensores y lógica probabilística |
| **Con aprendizaje** | Se adapta a incertidumbre y cambios ambientales |
| **Con autonomía** | Ejecuta decisiones sin depender de operadores humanos |

**Conclusión**: Los sistemas mecatrónicos inteligentes balancean estos tres elementos para crear agentes efectivos bajo incertidumbre, capaces de mejorar continuamente y actuar independientemente dentro de sus capacidades.

5. Defina cada tipo de agente en función de su **estructura** y dé un ejemplo de cada categoría.

Existen cinco tipos principales de agentes según la complejidad de su programa y estructura:

Agente reactivo simple: Su estructura se basa en reglas de condición-acción. Actúan únicamente según la percepción actual, sin considerar el historial de percepciones
. Requieren generalmente un entorno totalmente observable para tomar la decisión correcta
.
Ejemplo: Una aspiradora sin sensor de ubicación que simplemente aplica la regla: "si el cuadro actual está sucio, aspirar"

.
Agente reactivo basado en modelos: A diferencia del anterior, mantiene un estado interno que le permite seguir el rastro de aspectos del mundo que no son evidentes en la percepción actual
. Su estructura incorpora conocimiento sobre cómo evoluciona el entorno y cuáles son los efectos de sus propias acciones
.
Ejemplo: Un termostato inteligente que utiliza una ecuación para modelar la inercia térmica de una habitación, permitiéndole predecir cambios de temperatura y actuar de forma más eficiente

.
Agente basado en objetivos: Además del estado interno y el modelo del mundo, este agente cuenta con información sobre sus objetivos
. Su estructura le permite realizar procesos de búsqueda y planificación, proyectando qué sucederá si realiza una acción determinada para verificar si esta lo acerca a su meta
.
Ejemplo: Un agente de navegación por GPS para entrega de paquetes que analiza un grafo de calles para encontrar la ruta que garantice la entrega en el menor tiempo posible

.
Agente basado en utilidad: Utiliza una función de utilidad para elegir la mejor acción entre diversas alternativas que podrían llevar al objetivo
. Su estructura permite cuantificar el nivel de "felicidad" o éxito de un estado, lo que resulta fundamental cuando hay objetivos en conflicto o se debe equilibrar el rendimiento
.
Ejemplo: Un sistema de control de climatización en un edificio que debe decidir entre maximizar el confort térmico de los empleados o minimizar el costo de energía eléctrica

.
Agente que aprende: Su estructura se descompone en cuatro componentes conceptuales: el elemento de aprendizaje (realiza mejoras), el elemento de actuación (selecciona acciones externas), la crítica (proporciona realimentación basada en un estándar de actuación) y el generador de problemas (propone acciones exploratorias)
. Este proceso permite al agente mejorar su desempeño futuro basándose en su experiencia previa
.
Ejemplo: Un agente para jugar al Ti-Te-Ti (tres en línea) que, tras cada partida, recibe realimentación sobre si ganó o perdió y actualiza su conocimiento interno para refinar su estrategia en el siguiente juego
